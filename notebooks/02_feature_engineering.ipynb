```python
import pandas as pd

df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

df.info()
```

```python
# Create date and time features
df["InvoiceYear"] = df["InvoiceDate"].dt.year
df["InvoiceMonth"] = df["InvoiceDate"].dt.month
df["InvoiceDay"] = df["InvoiceDate"].dt.day
df["DayOfWeek"] = df["InvoiceDate"].dt.day_name()
df["InvoiceHour"] = df["InvoiceDate"].dt.hour

# Display the new columns
df[[
    "InvoiceDate",
    "InvoiceYear",
    "InvoiceMonth",
    "InvoiceDay",
    "DayOfWeek",
    "InvoiceHour"
]].head(10)
```

```python
# Customer-level feature engineering

customer_features = (
    df.groupby("Customer ID")
      .agg(
          TotalSpend=("TotalAmount", "sum"),
          TotalOrders=("Invoice", "nunique"),
          TotalProducts=("Quantity", "sum"),
          AverageOrderValue=("TotalAmount", "mean")
      )
      .reset_index()
)

customer_features.head()
```

```python
# First purchase date for each customer
first_purchase = (
    df.groupby("Customer ID")["InvoiceDate"]
      .min()
      .reset_index()
      .rename(columns={"InvoiceDate": "FirstPurchase"})
)

# Last purchase date for each customer
last_purchase = (
    df.groupby("Customer ID")["InvoiceDate"]
      .max()
      .reset_index()
      .rename(columns={"InvoiceDate": "LastPurchase"})
)

# Merge with customer features
customer_features = customer_features.merge(first_purchase, on="Customer ID")
customer_features = customer_features.merge(last_purchase, on="Customer ID")

# Calculate customer tenure (days)
customer_features["CustomerTenure"] = (
    customer_features["LastPurchase"] -
    customer_features["FirstPurchase"]
).dt.days

customer_features.head()
```

```python
# Avoid division by zero for customers with 0-day tenure
customer_features["TenureMonths"] = (
    customer_features["CustomerTenure"] / 30
).replace(0, 1)

# Purchase Frequency per Month
customer_features["PurchaseFrequencyPerMonth"] = (
    customer_features["TotalOrders"] /
    customer_features["TenureMonths"]
)

customer_features.head()
```

```python
# Average Spend per Product
customer_features["AverageSpendPerProduct"] = (
    customer_features["TotalSpend"] /
    customer_features["TotalProducts"]
)

customer_features.head()
```

```python
# Average Quantity per Order
customer_features["AverageQuantityPerOrder"] = (
    customer_features["TotalProducts"] /
    customer_features["TotalOrders"]
)

customer_features.head()
```

```python
# Latest date in the dataset
snapshot_date = df["InvoiceDate"].max() + pd.Timedelta(days=1)

# RFM features
rfm = (
    df.groupby("Customer ID")
      .agg({
          "InvoiceDate": lambda x: (snapshot_date - x.max()).days,
          "Invoice": "nunique",
          "TotalAmount": "sum"
      })
      .reset_index()
)

# Rename columns
rfm.columns = ["Customer ID", "Recency", "Frequency", "Monetary"]

rfm.head()
```

```python
# Merge customer features with RFM features

customer_dataset = customer_features.merge(
    rfm,
    on="Customer ID",
    how="left"
)

customer_dataset.head()
```

```python
# Remove duplicate column
customer_dataset = customer_dataset.drop(columns=["TotalSpend"])

customer_dataset.head()
```

```python
# Save feature engineered dataset

output_path = "../data/processed/customer_features.csv"

customer_dataset.to_csv(output_path, index=False)

print("Feature engineered dataset saved successfully!")
```
