```python
Q1 = df["Quantity"].quantile(0.25)
Q3 = df["Quantity"].quantile(0.75)

IQR = Q3 - Q1

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

quantity_outliers = df[
    (df["Quantity"] < lower_bound) |
    (df["Quantity"] > upper_bound)
]

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Quantity Outliers:", quantity_outliers.shape[0])
```

```python
quantity_outliers.head(20)
```

```python
df["Price"].describe()
```

```python
Q1_price = df["Price"].quantile(0.25)
Q3_price = df["Price"].quantile(0.75)

IQR_price = Q3_price - Q1_price

print("Q1 =", Q1_price)
print("Q3 =", Q3_price)
print("IQR =", IQR_price)
```

```python
lower_price = Q1_price - 1.5 * IQR_price
upper_price = Q3_price + 1.5 * IQR_price

print("Lower Bound:", lower_price)
print("Upper Bound:", upper_price)
```

```python
price_outliers = df[
    (df["Price"] < lower_price) |
    (df["Price"] > upper_price)
]

print("Number of Price Outliers:", price_outliers.shape[0])
```
